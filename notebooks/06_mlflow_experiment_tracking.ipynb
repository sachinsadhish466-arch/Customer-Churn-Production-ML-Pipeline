# MLflow Experiment Tracking

This notebook demonstrates experiment tracking for the Customer Churn Prediction project using MLflow.

The experiment will track:

- Model parameters
- Evaluation metrics
- Classification threshold
- Model artifacts
- Final XGBoost model

The objective is to make model development reproducible and easier to compare across experiments.

```python
mlflow.set_experiment(
    "Customer Churn Prediction"
)

print("MLflow experiment configured successfully.")
```

```python
import joblib
import json
import pandas as pd
import mlflow
import mlflow.sklearn

print("Libraries imported successfully.")
```

```python
best_xgb_pipeline = joblib.load(
    "../models/final_xgboost_pipeline.pkl"
)

best_threshold = joblib.load(
    "../models/churn_threshold.pkl"
)

with open(
    "../models/model_metadata.json",
    "r"
) as file:
    model_metadata = json.load(file)

print("Final model loaded successfully.")
print("Threshold:", best_threshold)
```

```python
mlflow.set_experiment(
    "Customer Churn Prediction"
)

print("MLflow experiment configured successfully.")
```

```python
import os
import mlflow

model_path = "../models/final_xgboost_pipeline.pkl"

print("Model exists:", os.path.exists(model_path))
print("Model path:", model_path)
```

```python
with mlflow.start_run(run_name="Final_XGBoost_Model"):

    mlflow.log_param(
        "model_type",
        model_metadata["model_type"]
    )

    mlflow.log_param(
        "classification_threshold",
        model_metadata["threshold"]
    )

    mlflow.log_metric(
        "cv_roc_auc",
        model_metadata["cv_roc_auc"]
    )

    mlflow.log_metric(
        "test_accuracy",
        model_metadata["test_accuracy"]
    )

    mlflow.log_metric(
        "test_precision",
        model_metadata["test_precision"]
    )

    mlflow.log_metric(
        "test_recall",
        model_metadata["test_recall"]
    )

    mlflow.log_metric(
        "test_f1",
        model_metadata["test_f1"]
    )

    mlflow.log_metric(
        "test_roc_auc",
        model_metadata["test_roc_auc"]
    )

    mlflow.log_artifact(
        model_path,
        artifact_path="model"
    )

    mlflow.log_artifact(
        "../models/churn_threshold.pkl",
        artifact_path="model"
    )

    mlflow.log_artifact(
        "../models/model_metadata.json",
        artifact_path="model"
    )

print("Final XGBoost model artifacts logged to MLflow successfully.")
```
