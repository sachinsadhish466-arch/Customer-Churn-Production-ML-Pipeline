## 10. Blank and Whitespace Value Check

Missing values are not always represented as `NaN`.

Some datasets may contain empty strings or whitespace characters.
Therefore, we check object/string columns for blank values before
performing data cleaning.

```python
blank_counts = {}

for column in df.select_dtypes(include="object").columns:
    blank_counts[column] = (df[column].astype(str).str.strip() == "").sum()

blank_counts = pd.Series(blank_counts)

blank_counts
```
